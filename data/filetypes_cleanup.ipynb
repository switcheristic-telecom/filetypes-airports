```python
# pandas import csv
import pandas as pd
import numpy as np


data = pd.read_csv('./raw/filetypes.csv',
                    keep_default_na=False, 
                    na_values=['NULL', 'NaN',''])

print(data.head(5))

```

```python
data.rename(columns={'Ext.': 'extension'}, inplace=True)
data.rename(columns={'Description': 'description'}, inplace=True)
data.rename(columns={'Used by': 'used_by'}, inplace=True)

# only keep these three columns
data = data[['extension', 'description', 'used_by']]

print(data.head(5))
```

```python
# save to csv
data.to_csv('./filetypes.csv', index=False)
```
