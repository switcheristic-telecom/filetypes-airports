```python
import csv

# load nirsoft-extensions-simplified.csv
with open('nirsoft-extensions-simplified.csv', 'r') as file:
    reader = csv.reader(file)
    extensions = list(reader)

# load airport_codes.csv
with open('airport_codes.csv', 'r') as file:
    reader = csv.reader(file)
    airports = list(reader)

```

```python
extensions = [x[0] for x in extensions]

for airport in airports: 
    iata = airport[9]
    if iata != "" and iata in extensions:
        print(iata)
```
