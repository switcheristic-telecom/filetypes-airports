```python
# pandas import csv
import pandas as pd
import numpy as np


data = pd.read_csv('./raw/airport_codes.tsv',
                    sep='\t',keep_default_na=False, 
                    na_values=['NULL', 'NaN',''])

print(data.head(5))

```

```python
coordinates = data["coordinates"]
lat =  coordinates.str.split(',').str.get(0)
lon =  coordinates.str.split(',').str.get(1)
data['latitude'] = lat
data['longitude'] = lon

# drop the coordinates column
data = data.drop('coordinates', axis=1)

print(data.head(5))
```

```python
# save the data to a new file

data.to_csv('./airport_codes.csv', index=False)
```

```python
# keep iata codes

columns = ['iata_code', 'name',  'type', 'iso_country', "municipality", 'latitude', 'longitude']

iata_data = data[columns]

# drop rows with missing iata codes
iata_data = iata_data.dropna(subset=['iata_code'])

print(iata_data.head(5))

print(iata_data.shape)

# save the data to a new file

iata_data.to_csv('./airport_codes_concise.csv', index=False)

```
